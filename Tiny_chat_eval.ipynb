```python
dialogues = [
    {"input": "hey how are you ?", "output": "i am feeling very happy today, thank you for asking."},
    {"input": "what is the meaning of love ?", "output": "love is a strong feeling that brings people together in many ways."},
    {"input": "do you like watching movies ?", "output": "i enjoy watching movies that have a wide range of interesting characters and plots."},
    {"input": "do you like reading books or watching movies better ?", "output": "i enjoy reading books because they take me to different worlds."},
    {"input": "tell me the best way to learn to read .", "output": "that sounds interesting and surprising. what kind of book do you enjoy reading?"},
    {"input": "i'm doing well , how about you ? .", "output": "i am feeling a bit bad today, thank you for asking."}
]
```

```python
inputs = [d['input'] for d in dialogues]
outputs = [d['output'] for d in dialogues]

rows = list(zip(inputs, outputs))

fig, ax = plt.subplots(figsize=(12, len(rows)*0.6))
ax.axis('off')

table = ax.table(
    cellText=rows,
    colLabels=['Input', 'Output'],
    cellLoc='left',
    loc='center',
    colWidths=[0.3, 0.7]
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.5)

plt.show()
```
