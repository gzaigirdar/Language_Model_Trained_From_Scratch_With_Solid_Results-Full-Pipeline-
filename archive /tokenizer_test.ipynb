```python
from transformers import AutoTokenizer, GPT2Tokenizer
tokenizer_path = 'Tokenizer/Saved_tokenizer/t5_Tokenizer'

t5 = AutoTokenizer.from_pretrained("t5-small",use_fast=False)
bpe = GPT2Tokenizer.from_pretrained("gpt2")
```

```python
t5.convert_ids_to_tokens([149, 33, 25, 58])
```

```python
text1 = 'Hello'
text2= ' hello hi '
bpe_encoded_text1 = bpe.encode(text1,add_special_tokens=False)
bpe_encoded_text2 = bpe.encode(text2,add_special_tokens=False)
bpe_decoded_text1 = bpe.decode(bpe_encoded_text1)
bpe_decoded_text2 = bpe.decode(bpe_encoded_text2)
t5_encoded_text1 = t5.encode(text1,add_special_tokens=False)
t5_encoded_text2 = t5.encode(text2,add_special_tokens=False)
t5_decoded_text1 = t5.decode(t5_encoded_text1)
t5_decoded_text2 = t5.decode(t5_encoded_text2)
```
